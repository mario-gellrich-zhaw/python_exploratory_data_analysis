# Solution: Interactive Graphics with Plotly — Swiss EV Charging Stations

```python
import pandas as pd
import plotly.express as px
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('../../data/ev_charging_stations.csv')
print('Shape:', df.shape)
```

### Task 1 — Histogram with marginal box plot

```python
fig = px.histogram(
    df, x='power_kw', nbins=40, marginal='box',
    color='power_type',
    title='Distribution of Charging Power (kW)',
    labels={'power_kw': 'Power (kW)'},
    template='plotly_white'
)
fig.show()
```

### Task 2 — Box plot: `power_kw` by `power_type`

```python
fig = px.box(
    df, x='power_type', y='power_kw', color='power_type',
    category_orders={'power_type': ['AC_1_PHASE', 'AC_3_PHASE', 'DC']},
    title='Charging Power by Power Type',
    labels={'power_kw': 'Power (kW)', 'power_type': 'Power Type'},
    template='plotly_white'
)
fig.show()
```

### Task 3 — Bar chart: top 15 cities

```python
top15 = df['city'].value_counts().head(15).reset_index()
top15.columns = ['city', 'count']

fig = px.bar(
    top15.sort_values('count'), x='count', y='city', orientation='h',
    title='Top 15 Cities by Number of EV Charging Stations',
    labels={'count': 'Number of Stations', 'city': 'City'},
    template='plotly_white'
)
fig.show()
```

### Task 4 — Scatter plot: `voltage` vs. `amperage`

```python
fig = px.scatter(
    df, x='voltage', y='amperage',
    color='power_type', size='power_kw',
    hover_data=['city', 'plug_type'],
    size_max=20, opacity=0.5,
    title='Voltage vs. Amperage (size = power_kw)',
    labels={'voltage': 'Voltage (V)', 'amperage': 'Amperage (A)'},
    template='plotly_white'
)
fig.show()
```

### Task 5 — Scatter plot matrix (SPLOM)

```python
num_cols = ['power_kw', 'amperage', 'voltage']
fig = px.scatter_matrix(
    df, dimensions=num_cols, color='power_type',
    opacity=0.4,
    title='Scatter Plot Matrix — EV Charging Stations',
    template='plotly_white'
)
fig.update_traces(diagonal_visible=False)
fig.show()
```

### Task 6 — Correlation heatmap

```python
corr = df[num_cols].corr(method='pearson').round(2)

fig = px.imshow(
    corr, text_auto=True, color_continuous_scale='RdBu_r',
    zmin=-1, zmax=1,
    title='Pearson Correlation Heatmap — EV Charging Stations',
    template='plotly_white'
)
fig.show()
```
