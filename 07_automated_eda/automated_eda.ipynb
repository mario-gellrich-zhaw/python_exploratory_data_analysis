## AI-Assisted EDA: Opportunities and Pitfalls

### Opportunities
- **Code generation**: generate EDA code (pandas, seaborn, plotly) from natural-language prompts
- **Plot suggestions**: AI can suggest suitable statistics and plots for the data types at hand
- **Explanation**: explain results, error messages, and unusual patterns
- **Notebook agents**: tools like GitHub Copilot (VS Code/Codespaces), Jupyter AI, and Colab Data Science Agent can run whole EDA workflows

### Pitfalls & Good Practice
- **Hallucinations**: AI can hallucinate column names, functions, or findings → verify every result against the actual data
- **Own responsibility**: keep the human in the loop — you are responsible for the correctness of the analysis
- **Data privacy**: never paste sensitive or personal data into public AI tools
- **Verification**: always run the generated code and check the output yourself

```python
# Quick manual profiling function as a lightweight alternative
import matplotlib.pyplot as plt
import seaborn as sns

def quick_profile(df: pd.DataFrame, max_cols: int = 10) -> None:
    """Print basic statistics and plot distributions for numerical columns."""
    num_cols = df.select_dtypes(include='number').columns[:max_cols]
    print('=== Quick Profile ===')
    print(f'Rows: {len(df):,}  |  Columns: {df.shape[1]}')
    print(f'Missing values:\n{df[num_cols].isnull().sum().to_string()}')
    print(f'\nDuplicates: {df.duplicated().sum()}')
    print(f'\nDescriptive statistics:')
    display(df[num_cols].describe().round(2))

    n = len(num_cols)
    fig, axes = plt.subplots(1, n, figsize=(4 * n, 3))
    if n == 1:
        axes = [axes]
    for ax, col in zip(axes, num_cols):
        sns.histplot(df[col].dropna(), bins=30, kde=True, ax=ax)
        ax.set_title(col)
        ax.set_xlabel('')
    fig.suptitle('Variable Distributions', fontsize=12, y=1.02)
    plt.tight_layout()
    plt.show()


quick_profile(df)
```
